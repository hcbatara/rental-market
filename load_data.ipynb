```python
import csv
import pandas as pd
from datetime import datetime
```

```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")
db = client["boston_housing"]
print("Connected!", client.list_database_names())
```

```python
# load permits
permits = pd.read_csv('permits.csv', dtype={'zip':str})
permits.head()
```

```python
permits.info()
```

```python
# drop na 
permits.dropna(inplace=True)
permits.info()
```

about 8k rows were dropped

```python
# convert money into float

permits['declared_valuation'] = permits['declared_valuation'].str.replace("$","").str.replace(",","")

permits.head()
```

```python
# Source - https://stackoverflow.com/a/26763793
# Posted by chrisb, modified by community. See post 'Timeline' for change history
# Retrieved 2026-04-12, License - CC BY-SA 4.0 - AI as well

# converting permit issue data into only the date and removing the exact time since i don't need that
permits['issued_date'] = pd.to_datetime(permits['issued_date'].str[:10], format='%Y-%m-%d', errors='coerce')
permits.head()
```

```python
# only keep the cols that i need
permits = permits[['permittypedescr', 'declared_valuation', 'issued_date', 'occupancytype', 'sq_feet', 'zip', 'y_latitude', 'x_longitude']]
permits.head()
```

```python
db['permits'].drop()
db['permits'].insert_many(permits.to_dict('records'))
print(f"Loaded {db['permits'].count_documents({})} permits into MongoDB")
```

***zillow rentals***

```python
rentals = pd.read_csv('rentals.csv', dtype={'RegionName':str})
rentals.head()
```

```python
# only boston rentals
rentals = rentals[rentals['City'] == 'Boston']
rentals.head()
```

```python
rentals.info()
```

```python
db["rent_prices"].drop()

for _, row in rentals.iterrows():
    ''' reshaping the date columns because right now, the each date column have prices in them.
    to make it better for mongodb i want them to be nested. used AI a lot in here to help through the loops '''
    
    rent_history = []
    for col in rentals.columns:
        if col.startswith("20"):
            val = row[col]
            if pd.notna(val):
                rent_history.append({
                    'date': datetime.strptime(col, "%Y-%m-%d"),
                    'rent': round(float(val), 2)
                })

    doc = {
    'zip': str(row['RegionName']).zfill(5),
    'city': row.get('City', 'Boston'),
    'metro': row.get('Metro', ''),
    'county': row.get('CountyName', ''),
    'rent_history': sorted(rent_history, key=lambda x: x['date']),
    'data_ovrtm': len(rent_history)
    }
    
    db["rent_prices"].insert_one(doc)

print(f"Loaded {db['rent_prices'].count_documents({})} ZIP codes into MongoDB")
```
